```python
import time
import torch
import numpy as np
import pandas as pd
import cv2
import matplotlib.pyplot as plt
from PIL import Image

from diffusers import (
    StableDiffusionPipeline,
    PNDMScheduler, DDIMScheduler, LMSDiscreteScheduler,
    EulerDiscreteScheduler, EulerAncestralDiscreteScheduler,
    DPMSolverMultistepScheduler, UniPCMultistepScheduler,
)
from transformers import CLIPModel, CLIPProcessor

device = "cuda" if torch.cuda.is_available() else "cpu"      # use GPU if available
dtype = torch.float16 if device == "cuda" else torch.float32  # half precision on GPU
print("Device:", device, "| dtype:", dtype)
```

```python
MODEL_ID = "stable-diffusion-v1-5/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(MODEL_ID, torch_dtype=dtype)
pipe = pipe.to(device)
pipe.enable_attention_slicing()          # reduces GPU memory use

BASE_CONFIG = pipe.scheduler.config      # keep original config to build other schedulers
print("Default scheduler:", pipe.scheduler.__class__.__name__)
```

```python
TEST_PROMPT = "a majestic lion standing on a rocky cliff at sunrise, golden hour lighting, highly detailed fur, 4k, photorealistic"
NEGATIVE = "blurry, low quality, distorted, deformed, extra limbs, watermark, text, cropped"

SCHEDULERS = {
    "PNDM (default)":  lambda c: PNDMScheduler.from_config(c),
    "DDIM":            lambda c: DDIMScheduler.from_config(c),
    "LMS":             lambda c: LMSDiscreteScheduler.from_config(c),
    "Euler":           lambda c: EulerDiscreteScheduler.from_config(c),
    "Euler A":         lambda c: EulerAncestralDiscreteScheduler.from_config(c),
    "DPM++ 2M":        lambda c: DPMSolverMultistepScheduler.from_config(c),
    "DPM++ 2M Karras": lambda c: DPMSolverMultistepScheduler.from_config(c, use_karras_sigmas=True),
    "UniPC":           lambda c: UniPCMultistepScheduler.from_config(c),
}

def set_scheduler(name):
    pipe.scheduler = SCHEDULERS[name](BASE_CONFIG)   # swap sampler without reloading model

def generate(prompt, negative_prompt=None, steps=30, guidance=7.5,
             scheduler="DPM++ 2M", seed=42, height=512, width=512):
    set_scheduler(scheduler)
    gen = torch.Generator("cpu").manual_seed(seed)   # fixed seed = same starting noise
    if device == "cuda": torch.cuda.synchronize()
    start = time.time()
    out = pipe(prompt=prompt, negative_prompt=negative_prompt,
               num_inference_steps=steps, guidance_scale=guidance,
               generator=gen, height=height, width=width)
    if device == "cuda": torch.cuda.synchronize()
    return out.images[0], time.time() - start

def show_grid(images, titles, ncols=4, size=3.2, suptitle=None):
    nrows = int(np.ceil(len(images) / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(size * ncols, size * nrows + 0.5))
    axes = np.array(axes).reshape(-1)
    for ax in axes: ax.axis("off")
    for ax, im, t in zip(axes, images, titles):
        ax.imshow(im); ax.set_title(t, fontsize=9)
    if suptitle: fig.suptitle(suptitle, fontsize=13)
    plt.tight_layout(); plt.show()
```

```python
clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
clip_proc = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

@torch.no_grad()
def clip_score(image, prompt):                      # image-text similarity (higher = better match)
    inputs = clip_proc(text=[prompt], images=image, return_tensors="pt",
                       padding=True, truncation=True, max_length=77).to(device)
    return clip_model(**inputs).logits_per_image.item()

def sharpness(image):                               # variance of Laplacian (higher = more detail)
    gray = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var()

def mse(img_a, img_b):                              # pixel difference (lower = closer)
    a = np.asarray(img_a, dtype=np.float32) / 255.0
    b = np.asarray(img_b, dtype=np.float32) / 255.0
    return float(((a - b) ** 2).mean())

results = []   # every experiment appends rows here
```

```python
prompt_pairs = [
    ("a lion",   "a majestic lion standing on a rocky cliff at sunrise, golden hour lighting, highly detailed fur, 4k, photorealistic"),
    ("a castle", "a fairytale castle on a mountain above the clouds, dramatic sky, volumetric light, intricate architecture, digital painting, 8k"),
    ("a robot",  "a friendly retro robot reading a book in a cozy library, warm lamp light, detailed, cinematic, shallow depth of field"),
]

images, titles = [], []
for basic, engineered in prompt_pairs:
    img_b, _ = generate(basic, steps=30, guidance=7.5)
    img_e, _ = generate(engineered, negative_prompt=NEGATIVE, steps=30, guidance=7.5)
    images += [img_b, img_e]
    titles += [f"Basic: {basic}", "Engineered + negative prompt"]
    results.append(dict(experiment="prompt", setting=f"basic: {basic}",
                        CLIP=clip_score(img_b, engineered), sharpness=sharpness(img_b)))
    results.append(dict(experiment="prompt", setting=f"engineered: {basic}",
                        CLIP=clip_score(img_e, engineered), sharpness=sharpness(img_e)))

show_grid(images, titles, ncols=2, size=4.2, suptitle="Prompt engineering: basic vs detailed prompt")
```

```python
step_list = [5, 10, 15, 20, 30, 50, 100]
step_imgs, step_times = {}, {}
for s in step_list:
    step_imgs[s], step_times[s] = generate(TEST_PROMPT, NEGATIVE, steps=s,
                                           guidance=7.5, scheduler="DPM++ 2M")

ref_img = step_imgs[100]   # reference for convergence (MSE)
for s in step_list:
    results.append(dict(experiment="steps", setting=s, time_s=step_times[s],
                        CLIP=clip_score(step_imgs[s], TEST_PROMPT),
                        sharpness=sharpness(step_imgs[s]),
                        MSE_vs_ref=mse(step_imgs[s], ref_img)))

show_grid([step_imgs[s] for s in step_list],
          [f"{s} steps ({step_times[s]:.1f}s)" for s in step_list],
          ncols=4, suptitle="Effect of inference steps (guidance=7.5, DPM++ 2M)")

df = pd.DataFrame([r for r in results if r["experiment"] == "steps"])
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
ax[0].plot(df.setting, df.time_s, "o-");  ax[0].set_title("Time vs steps"); ax[0].set_xlabel("steps"); ax[0].set_ylabel("seconds")
ax[1].plot(df.setting[:-1], df.MSE_vs_ref[:-1], "o-r"); ax[1].set_title("MSE vs 100-step image (lower = converged)"); ax[1].set_xlabel("steps")
ax[2].plot(df.setting, df.CLIP, "o-g");   ax[2].set_title("CLIP score vs steps"); ax[2].set_xlabel("steps")
for a in ax: a.grid(alpha=0.3)
plt.tight_layout(); plt.show()
df
```

```python
guidance_list = [1, 3, 5, 7.5, 10, 15, 20]
g_imgs = {}
for g in guidance_list:
    g_imgs[g], t = generate(TEST_PROMPT, NEGATIVE, steps=30, guidance=g, scheduler="DPM++ 2M")
    results.append(dict(experiment="guidance", setting=g, time_s=t,
                        CLIP=clip_score(g_imgs[g], TEST_PROMPT), sharpness=sharpness(g_imgs[g])))

show_grid([g_imgs[g] for g in guidance_list], [f"guidance = {g}" for g in guidance_list],
          ncols=4, suptitle="Effect of guidance scale (steps=30, DPM++ 2M)")

dg = pd.DataFrame([r for r in results if r["experiment"] == "guidance"])
fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
ax[0].plot(dg.setting, dg.CLIP, "o-g");      ax[0].set_title("CLIP score vs guidance");  ax[0].set_xlabel("guidance scale")
ax[1].plot(dg.setting, dg.sharpness, "o-m"); ax[1].set_title("Sharpness vs guidance");   ax[1].set_xlabel("guidance scale")
for a in ax: a.grid(alpha=0.3)
plt.tight_layout(); plt.show()
dg
```

```python
sched_steps = [5, 10, 20, 30]
s_imgs, s_times = {}, {}
for name in SCHEDULERS:
    for s in sched_steps:
        s_imgs[(name, s)], s_times[(name, s)] = generate(TEST_PROMPT, NEGATIVE, steps=s,
                                                         guidance=7.5, scheduler=name)
        results.append(dict(experiment="scheduler", setting=f"{name} @ {s}",
                            time_s=s_times[(name, s)],
                            CLIP=clip_score(s_imgs[(name, s)], TEST_PROMPT),
                            sharpness=sharpness(s_imgs[(name, s)]),
                            MSE_vs_ref=mse(s_imgs[(name, s)], ref_img)))

for s in [20, 10]:
    show_grid([s_imgs[(n, s)] for n in SCHEDULERS],
              [f"{n} ({s_times[(n, s)]:.1f}s)" for n in SCHEDULERS],
              ncols=4, suptitle=f"Samplers compared at {s} steps (guidance=7.5)")

dsch = pd.DataFrame([r for r in results if r["experiment"] == "scheduler"])
dsch["scheduler"] = dsch.setting.str.split(" @ ").str[0]
dsch["steps"] = dsch.setting.str.split(" @ ").str[1].astype(int)

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))
for name, g in dsch.groupby("scheduler", sort=False):
    ax[0].plot(g.steps, g.MSE_vs_ref, "o-", label=name)
    ax[1].plot(g.steps, g.CLIP, "o-", label=name)
ax[0].set_title("Convergence: MSE vs reference (lower = better)"); ax[0].set_xlabel("steps")
ax[1].set_title("CLIP score vs steps"); ax[1].set_xlabel("steps")
for a in ax: a.grid(alpha=0.3)
ax[0].legend(fontsize=8); plt.tight_layout(); plt.show()

dsch.pivot(index="scheduler", columns="steps", values="CLIP").round(2)
```

```python
grid_steps = [10, 20, 40]
grid_guid = [3, 7.5, 12]
gimgs, gtitles = [], []
clip_grid = np.zeros((len(grid_steps), len(grid_guid)))

for i, s in enumerate(grid_steps):
    for j, g in enumerate(grid_guid):
        img, _ = generate(TEST_PROMPT, NEGATIVE, steps=s, guidance=g, scheduler="DPM++ 2M")
        clip_grid[i, j] = clip_score(img, TEST_PROMPT)
        gimgs.append(img); gtitles.append(f"steps={s}, guidance={g}")

show_grid(gimgs, gtitles, ncols=len(grid_guid), size=4, suptitle="Steps x guidance grid")

plt.figure(figsize=(5, 4))
plt.imshow(clip_grid, cmap="viridis")
plt.xticks(range(len(grid_guid)), grid_guid); plt.yticks(range(len(grid_steps)), grid_steps)
plt.xlabel("guidance scale"); plt.ylabel("inference steps"); plt.title("CLIP score heatmap"); plt.colorbar()
for i in range(len(grid_steps)):
    for j in range(len(grid_guid)):
        plt.text(j, i, f"{clip_grid[i, j]:.1f}", ha="center", va="center", color="white")
plt.show()
```
